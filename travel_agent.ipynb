```python
import os
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.callbacks import StreamingStdOutCallbackHandler
from dotenv import load_dotenv

load_dotenv()
```

```python
summary_prompt = ChatPromptTemplate.from_template("""Summarize the following conversation
breifly keeping only the important information and the context of the conversation
Conversation so far:
{history}
""")

def summary_agent():
    global memory

    history_text = "/n".join(memory)

    summary_llm = ChatOpenAI(
        model="gpt-4o-mini",
        temperature=0.7)

    summary_chain = summary_prompt | summary_llm

    summary=summary_chain.invoke({"history": history_text}).content

    memory = [f"Summary: {summary}"]

    print(f"Summary of the conversation so far: {summary}" )
    

```
